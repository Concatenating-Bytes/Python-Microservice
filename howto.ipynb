#### For Development (Auto-Reloads on Save)

<p>Use this command to run the madafaka python server. It runs on localhost and auto-restarts on ever ctrl+s</p>

```python
uvicorn app:app --reload
```

#### For Integration (in production)

<p>when running the server to serve real-time requests</p>

```python
uvicorn app:app --host 0.0.0.0 --port 8000
```

#### For endpoint testings

<p>To test the endpoints defined in app.py, use a injection testing file like test.py and feed the endpoints directly</p>

```python
python test.py
```
